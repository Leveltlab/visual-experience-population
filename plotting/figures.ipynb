# Flagship figure panels

Export B–D, F–H and B supplements from the included deconvolved results. Each panel is saved independently as editable SVG, outlined SVG, PDF and PNG. This notebook does not refit analyses. See the repository README for recomputation and the statistical caption.

D/H use descriptive group-mean PCA; F and H use different contrast pools. All paths below are portable.

```python
from pathlib import Path
import os

ROOT = Path.cwd().resolve()
if ROOT.name == "plotting":
    ROOT = ROOT.parent
assert (ROOT / "protocol.json").is_file(), "Start Jupyter in the repository root or plotting folder"
RESULTS = Path(
    os.environ.get("VEP_FIGURE_DATA", ROOT / "plotting" / "source_data")
)  # Or ROOT / "results" / "spikes"
EXPORTS = Path(os.environ.get("VEP_EXPORT_DIR", ROOT / "plotting" / "exports"))
os.environ["VEP_RESULTS_DIR"] = str(RESULTS)
from plotting.export import export_panel
from IPython.display import display, Image
```
